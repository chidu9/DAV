```python
import numpy as np
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

print(a)
```

```python
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a ** 2)
```

```python
print(np.sum(a))
print(np.mean(a))
print(np.min(a))
print(np.max(a))

print(np.sqrt(a))
print(np.std(a))
```

```python
print(a[0])
print(a[-1])
print(a[1:4])
```
