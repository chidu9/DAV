```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.DataFrame({
    "MARKS":[10,20,None,50,100],
    "ATTENDENCE":[60,50,22,None,11],
    "DEPARTMENT":["CSE","ECE","CSE","ECE",None]
})
```

```python
#mean imputation
df["MARKS"]=df["MARKS"].fillna(df["MARKS"].mean())

#median imputation
df["ATTENDENCE"]=df["ATTENDENCE"].fillna(df["ATTENDENCE"].median())

#forward fill (fills the none with previous or above values)
df["ATTENDENCE"]=df["ATTENDENCE"].ffill()

#fill missing categorial value using mode
df["DEPARTMENT"]=df["DEPARTMENT"].fillna(df["DEPARTMENT"].mode()[0])

#label encoding (converts categorical text into numbers)
encoder=LabelEncoder()
df["DEPARTMENT"]=encoder.fit_transform(df["DEPARTMENT"])

print(df)
```
